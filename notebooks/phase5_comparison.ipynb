# Phase 5: Four-Way Comparison Plots

Pull WandB data from `phase5-comparison` group and produce thesis-quality figures.
- Figure 1: Regret vs Updates for all four methods (ACCEL, CMA-ES, NS-ES, SV-CMA-ES)
- Figure 2: Ablation placeholder (Phase 6)

```python
import wandb
import pandas as pd
import matplotlib
matplotlib.use("Agg")  # Non-interactive backend for server
import matplotlib.pyplot as plt

# Pull runs from WandB
api = wandb.Api()
runs = api.runs("es-accel", filters={"group": "phase5-comparison"})

METHOD_ORDER = ["accel-baseline", "cma-es", "ns-es", "sv-cma-es"]
COLORS = {
    "accel-baseline": "#555555",
    "cma-es": "#1f77b4",
    "ns-es": "#ff7f0e",
    "sv-cma-es": "#2ca02c",
}
LABELS = {
    "accel-baseline": "ACCEL (baseline)",
    "cma-es": "CMA-ES",
    "ns-es": "NS-ES",
    "sv-cma-es": "SV-CMA-ES (ours)",
}

dfs = {}
for run in runs:
    df = run.history(keys=["update", "mean_buffer_score"], pandas=True)
    df = df.sort_values("update").dropna(subset=["mean_buffer_score"])
    dfs[run.name] = df
    print(f"Loaded {run.name}: {len(df)} data points")

print(f"\nRuns found: {list(dfs.keys())}")
```

```python
WINDOW = 50  # Rolling mean window

fig, ax = plt.subplots(figsize=(8, 5))

for name in METHOD_ORDER:
    if name not in dfs:
        print(f"Warning: {name} not found in WandB data, skipping")
        continue
    df = dfs[name]
    smoothed = df["mean_buffer_score"].rolling(WINDOW, min_periods=1).mean()
    ax.plot(
        df["update"], smoothed,
        label=LABELS.get(name, name),
        color=COLORS.get(name, None),
        linewidth=1.5,
    )

ax.set_xlabel("Training Updates", fontsize=12)
ax.set_ylabel("Mean Buffer Score (regret proxy)", fontsize=12)
ax.set_title("Phase 5: Four-Method Comparison (seed=42)", fontsize=13)
ax.legend(fontsize=10, loc="upper left")
ax.grid(True, alpha=0.3)
ax.tick_params(labelsize=10)
fig.tight_layout()

import os
os.makedirs("figures", exist_ok=True)
fig.savefig("figures/phase5_comparison.pdf", bbox_inches="tight", dpi=300)
fig.savefig("figures/phase5_comparison.png", bbox_inches="tight", dpi=300)
print("Saved: figures/phase5_comparison.pdf and .png")
plt.show()
```
